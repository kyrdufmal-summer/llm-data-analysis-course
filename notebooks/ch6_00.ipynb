```python
#완료된 주문건 조회 (orders) 
completed_orders = orders[
    orders["order_status"] == "completed"
]

print(completed_orders["order_status"].value_counts())
completed_orders.head()
```

```python
# order_items에 완료된 주문 정보 연결
items_with_orders = order_items.merge(
    completed_orders[
        ["order_id", "customer_id", "order_date", "order_status"]
    ],
    on="order_id",
    how="inner",
    validate="many_to_one",
    indicator=True,
)

items_with_orders.head()
```

```python
# 고객별 평균 주문 건수?
completed_item_with_orders = items_with_orders[
    items_with_orders["order_status"] == "completed"
]

print(completed_item_with_orders["order_status"].value_counts())
```

```python
#고객별 완료 주문 건수?
customer_completed_orders = completed_item_with_orders.groupby("customer_id")["order_id"].nunique()

customer_completed_orders.head()
```
