```python
from pathlib import Path
import sys

current = Path.cwd().resolve()

while not (current / "course_utils").is_dir():
    current = current.parent

sys.path.append(str(current))

print("import 기준 경로:", current)
# 현재 폴더를 확인한다.
current = Path.cwd().resolve() #resolve는 전체경로 나오게
project_root = current
print("현재 위치:",current)

# 현재 폴더 안에 data 폴더가 있는지 확인한다. (2)
while True:
    is_data = current / "data"
    print("현재 폴더에 data라는 하위 폴더가 존재하는가? ",is_data.is_dir())
    # 만약 data 폴더가 있다면
    if (current / "data").is_dir():
    #    현재 폴더를 프로젝트 루트로 결정한다.
        project_root = current
    #    탐색을 종료한다.
        break

    # 만약 data 폴더가 없다면
    # 부모 폴더로 한 단계 올라간다.
    current = current.parent
# 다시 data 폴더가 있는지 확인한다.
# 이 과정을 반복한다.
# 더 이상 위로 올라갈 수 없는데도
# 데이터 폴더를 찾지 못했다면

# 오류를 발생시킨다.

print("프로젝트 루트:", project_root)
```
