STEP 3. 전처리 전 상태 기록

```python
raw_data = {
    "customers": customers,
    "products": products,
    "orders": orders,
    "order_items": order_items,
}

for name, df in raw_data.items():
    print(f"#{name}")
    print("shape:", df.shape)
    print("missing:", int(df.isna().sum().sum()))
    print("duplicated rows:", int(df.duplicated().sum()))
    print()
```
