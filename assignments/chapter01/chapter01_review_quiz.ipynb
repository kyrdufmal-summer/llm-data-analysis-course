### 컬럼명 소문자 변환 전 생각

- 반복할 대상: `columns`에 들어 있는 컬럼명들
- 새 컬럼명을 저장할 자료구조: 빈 리스트 `lower_list`
- 문자열 하나를 소문자로 바꾸는 방법: `.lower()` 메서드를 사용한다.
- 바뀐 문자열을 어디에 추가할 것인가: 소문자로 바꾼 컬럼명을 `lower_list`에 하나씩 추가한다.
- 반복 처리 방법: `for`문으로 `columns`의 컬럼명을 하나씩 확인한다.
- 최종 결과: 모든 컬럼명이 소문자로 바뀐 `lower_list`를 출력한다.
- 아직 헷갈리는 부분:
  - `for`문으로 컬럼명을 하나씩 꺼내는 과정
  - `.lower()`로 바꾼 값을 리스트에 추가하는 과정

```python
import pandas as pd
url = "https://raw.githubusercontent.com/GilbertMoon/llm-data-analysis-course/main/data/raw/customers.csv"
df = pd.read_csv(url)

#이미 csv파일의 컬럼명이 존재하므로 필요없는 작업이긴 함.(전부 주석처리해도 진행가능.)
df.columns= [
"Customer_ID",
"Name",
"Gender",
"Age",
"City",
"Signup_Date"
]
#한 행을 하나의 딕셔너리로 만들어라. orient="records" -> 
# {
# "cumsomer_id :1,
# "name":"김수민",
# "gender":f
#}
records= df.to_dict(orient="records")
print("#records:",records )
print("================")
columns= list(df.columns)
print(df.head(10))
print(df.shape)
print("전체 크기:", df.shape)

print("\n처음 5개")
print(df.head())

print("\n마지막 5개")
print(df.tail())
```

```python
#요구사항1—전체데이터를순회한다
#records안에있는모든고객을한명씩확인합니다.

# 가장 오래된 가입일 (가장 첫번째 고객의 가입일을 기준으로 잡는다.)
oldest_signup_date = records[0]["Signup_Date"]

# 가장 최근 가입일 (가장 첫번째 고객의 가입일을 기준으로 잡는다.)
latest_signup_date = records[0]["Signup_Date"]

#요구사항1: 전체데이터를 순회한다.
for record in records:
    name = record["Name"]
    signup_Date = record["Signup_Date"]

#요구사항2—가입일의 최댓값과 최솟값을 구한다
#현재 고객의 가입일이 latest_signup_date보다 더 최근 이라면, 현재 고객의 가입일로 바꾼다.
    if signup_Date > latest_signup_date:
        latest_signup_date =signup_Date
#현재 고객의 가입일이 oldest_signup_date보다 더 예전 가입 이라면, 현재 고객의 가입일로 바꾼다.
    if  oldest_signup_date > signup_Date:
        oldest_signup_date = signup_Date

print(f"가장최근 가입일: {latest_signup_date}")
print(f"가장오랜 가입일: {oldest_signup_date}")

         

```

```python
#요구사항3—나이의합계와평균을구한다

# 가장 연장자
oldest_age = records[0]["Age"]

# 가장 어린 사람
youngest_age = records[0]["Age"]

#사람들의 나이 합계
total_Age = 0

#사람 수 변수
person_count = 0

for record in records:
    age = record["Age"]
    total_Age += age
    person_count += 1

age_Avg = total_Age/person_count

print(f"전체나이 합계: {total_Age}")
print(f"전체나이 평균: {age_Avg}")
    
```

나의 답변:

반복문이 한 번 실행될 때마다 현재 고객의 가입일을 기존 날짜와 비교한다. 더 최근의 날짜이면 최근 가입일을 바꾸고, 더 오래된 날짜이면 가장 오래된 가입일을 바꾼다. 동시에 현재 고객의 나이를 나이 합계에 더하고 사람 수를 1씩 증가시킨다.

## 5. pandas의 `sum()`, `mean()`, `max()`, `min()`을 사용하지 않고 직접 구현해 보니 무엇을 이해하게 되었는가?

나의 답변:

내부적으로는 데이터를 하나씩 확인하면서 값을 비교하거나 더하는 과정이 필요하다는 것을 이해하게 되었다. `sum()`은 값을 누적하고, `max()`와 `min()`은 기준값과 비교하며, 평균은 합계와 사람 수를 이용해 계산한다는 원리를 알게 되었다.

## 6. 아직 이해가 안 되는 부분은 무엇인가?

나의 답변:

첫 번째 데이터를 기준값으로 사용하는 방법과 데이터가 비어 있을 때 예외를 처리하는 방법이 더 궁금하다.

## 7. 같은 문제를 다시 푼다면 가장 먼저 무엇을 할 것인가?

나의 답변:

먼저 문제에서 구해야 하는 값과 반복해서 처리해야 하는 데이터를 구분할 것이다. 그다음 필요한 변수의 역할과 처음 값을 정하고, 반복문 안에서 어떤 값을 더하거나 비교할지 작업 계획부터 작성할 것이다.

## 8. AI를 사용했다면, AI에게 묻지 않고 스스로 해결할 수 있었던 질문은 무엇이었는가?

나의 답변:

`records`가 리스트이고 고객 한 명의 데이터가 딕셔너리라는 점, `columns`를 반복하면서 `.lower()`로 소문자 변환을 하는 부분은 스스로 해결할 수 있었다. 또한 나이 합계와 사람 수를 반복문 안에서 누적하고 평균을 반복문이 끝난 뒤 계산하는 구조도 스스로 생각할 수 있었다.
