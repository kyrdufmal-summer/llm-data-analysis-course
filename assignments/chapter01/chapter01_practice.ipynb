### 6. 학생 성적 문제와 비교

학생 성적 문제와 전체 구조는 같다.

- 학생 문제: 점수가 80점 이상인지 판단
- 쇼핑몰 문제: 주문 금액이 50,000원 이상인지 판단

즉, 데이터와 조건의 기준값만 달라지고  
`여러 개의 데이터를 한 개씩 반복 → 조건 판단 → 결과 출력` 구조는 동일하다.


```python
customers = [
    {"name":"민수","amount":45000},
    {"name":"지영","amount":72000},
    {"name":"현우","amount":53000},
    {"name":"수진","amount":49000},
]

for customer in customers:
    name = customer["name"]
    amount = customer["amount"]
    if amount >= 50000:
        result = "무료배송"
    else:
        result = "배송비 있음"    

print(f"{name}님의 주문금액은: {amount} 입니다, 배송비는 {result}입니다.")
```

5. 100,000원 이상이면 `우수회원`으로 저장한다.
6. 그렇지 않으면 `일반회원`으로 저장한다.
