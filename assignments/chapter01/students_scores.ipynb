```python
#7. 쇼핑몰 주문의 배송 등급 결정하기


customers = [
    {"name": "민수" ,"order_price": 49000, "grade" : "일반"},
    {"name":"지영","order_price":50000,"grade":"VIP"},
    {"name":"현우","order_price":99000,"grade":"일반"},
    {"name":"수빈","order_price":100000,"grade":"VIP"},
    {"name":"원빈","order_price":120000,"grade":"일반"},
]

#모든 주문을 하나씩 확인하고 다음 기준으로 배송 결과를 정합니다.
for customer in customers:
    order_price = customer["order_price"]
    name = customer["name"]
    grade = customer["grade"]
    ship_type =""

    if order_price >= 100000 and grade =="VIP":
        ship_type ="VIP 유선배송"
        # print(f"{name}님의 배송방법은 VIP 우선배송")
    elif order_price >= 50000 and grade =="VIP":
        ship_type ="무료배송"
        # print(f"{name}님은 무료배송")
    else :
        ship_type ="일반배송"
        # print(f"{name}님은 일반배송")    

    print(f"{name}의 배송결과는{ship_type} 입니다. 주문금액: {order_price}원, 회원등급:{grade}")

# • 주문금액이 100,000원 이상이면서 회원등급이 VIP이면 VIP 우선배송

# • 그렇지 않더라도 주문금액이 50,000원 이상이거나 회원등급이 VIP이면 무료배송

# • 나머지는 일반배송

# • 모든 고객의 고객명, 주문금액, 회원등급, 배송 결과를 확인할 수 있어야 합니다.
```
